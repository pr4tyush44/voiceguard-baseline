# VoiceGuard small anti-spoofing baseline

This notebook runs the corrected training pipeline in this folder. It trains only on ASVspoof `train`, chooses its operating threshold on `dev`, and reports a held-out `eval` estimate. It does **not** prove performance on real calls or every voice-cloning system.

```python
# In a fresh environment, install the requirements once:
# %pip install -r requirements.txt

from pathlib import Path
import subprocess
import sys

# Change this to the extracted ASVspoof 2019 Logical Access dataset.
DATASET_ROOT = Path(r"D:\\datasets\\ASVspoof2019\\LA\\LA")
MODEL_OUT = Path("voiceguard_baseline.joblib")
REPORT_OUT = Path("training_report.json")
assert DATASET_ROOT.exists(), f"Set DATASET_ROOT first: {DATASET_ROOT}"
```

```python
command = [
    sys.executable, "train_baseline.py",
    "--dataset-root", str(DATASET_ROOT),
    "--model-out", str(MODEL_OUT),
    "--report-out", str(REPORT_OUT),
]
subprocess.run(command, check=True)
```

```python
import json
report = json.loads(REPORT_OUT.read_text(encoding="utf-8"))
print(json.dumps({
    "dev_eer": report["dev_eer"],
    "eval": report["eval"],
}, indent=2))
print("Read the full report before claiming any model performance.")
```
